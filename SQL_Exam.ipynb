### Question 1

Who won the Oscar for “Actor in a Leading Role” in  2015?

(Hint: The winner is indicated as '1.0'.)

**Options:** 

  - Micheal Fassbender
  - Natalie Portman
  - Leonardo DiCaprio

### Question 5

How many movies are there that are both in the "Thriller" genre and contain the word “love” anywhere in the keywords?


**Options:**
 - 48
 - 38
 - 14
 - 1

```python
%%sql
SELECT
    *
FROM
    movies as m
JOIN
    keywords as k
JOIN
    keywordmap as km
JOIN
    genres as g
JOIN
    genremap as gm
ON
    m.movie_id = km.movie_id
AND
    k.keyword_id = km.keyword_id
AND
    m.movie_id = gm.movie_id
AND
    g.genre_id = gm.genre_id
WHERE
    g.genre_name = "Thriller"
AND
    k.keyword_name LIKE "%love%";
```

### Question 7

How many unique characters has "Vin Diesel" played so far in the database?

**Options:**

```python
%%sql
SELECT
    a.actor_id,
    a.actor_name,
    COUNT(c.characters)
FROM
    actors as a
JOIN
    casts as c
ON
    a.actor_id = c.actor_id
WHERE
    actor_name = "Vin Diesel";
```

### Question 8

What are the genres of the movie “The Royal Tenenbaums”?


**Options:**
 - Action, Romance
 - Drama, Comedy
 - Crime, Thriller
 - Drama, Romance

```python
%%sql
SELECT
    *
FROM
    genres as g
JOIN
    movies as m
JOIN
    genremap as gm
ON
    m.movie_id = gm.movie_id
AND
    g.genre_id = gm.genre_id
WHERE
    m.title = "The Royal Tenenbaums";
```

### Question 9

What are the three production companies that have the highest movie popularity score on average, as recorded within the database?


**Options:**

 - MCL Films S.A., Turner Pictures, and George Stevens Productions
 - The Donners' Company, Bulletproof Cupid, and Kinberg Genre
 - Bulletproof Cupid, The Donners' Company, and MCL Films S.A
 - B.Sting Entertainment, Illumination Pictures, and Aztec Musique

```python
%%sql
SELECT
    p.production_company_name,
    AVG(m.popularity) as Average_popularity
FROM
    movies as m
JOIN
    productioncompanymap as pm
ON
    m.movie_id = pm.movie_id
JOIN
    productioncompanies as p
ON
    p.production_company_id = pm.production_company_id
GROUP BY
    p.production_company_name
ORDER BY
    Average_popularity DESC
LIMIT
    3;
```

### Question 10

How many female actors (i.e. gender = 1) have a name that starts with the letter "N"?


**Options:**

 - 0
 - 355
 - 7335
 - 1949

```python
%%sql
SELECT
    actor_name,
    COUNT(DISTINCT(actor_id)) as Num_female
FROM
    actors
WHERE
    gender = "1"
AND
    actor_name LIKE "N%";
```

### Question 11

Which genre has, on average, the lowest movie popularity score? 


**Options:**

 - Science Fiction
 - Animation
 - Documentary
 - Foreign

```python
%%sql
SELECT
    m.movie_id,
    m.title,
    g.genre_name,
    AVG(m.popularity)
FROM
    movies as m
JOIN
    genremap as gm
JOIN
    genres as g
ON
    m.movie_id = gm.movie_id
AND
    gm.genre_id = g.genre_id
GROUP BY
    g.genre_name
ORDER BY
    AVG(m.popularity) ASC;
```

### Question 12

Which award category has the highest number of actor nominations (actors can be male or female)? (Hint: `Oscars.name` contains both actors' names and film names.)

**Options:**

- Special Achievement Award
- Actor in a Supporting Role
- Actress in a Supporting Role
- Best Picture



```python
%%sql
SELECT
    award,
    COUNT(award) as Num_of_nominations
FROM
    oscars
WHERE
    award IN ("Special Achievement Award", "Actor in a Supporting Role", "Actress in a Supporting Role", "Best Picture")
GROUP BY
    award
ORDER BY
    award;
```
